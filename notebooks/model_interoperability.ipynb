```python
import joblib

xgb_model = joblib.load("../outputs/xgb_fraud_model.pkl")
```

```python
import pandas as pd
#importing test set
X_fraud_test = pd.read_csv("../data/processed/test_data_X.csv")
```

```python
import shap
# Initialize the SHAP explainer
explainer = shap.TreeExplainer(xgb_model)

# Compute SHAP values for the test set
shap_values = explainer.shap_values(X_fraud_test)  # fraud dataset
```

```python
shap.summary_plot(shap_values, X_fraud_test)
```
